## Plotting Bairnsdale demand Australia Day 2009 - 2015

```python
import matplotlib.pyplot as plt

ausday_data = {}

for f in sorted(clean_dir.glob("AusNet_VIC_*_cleaned.csv")):
    df = pd.read_csv(f, parse_dates=["StartDeliveryTime"], index_col="StartDeliveryTime")

    if first_substation not in df.columns:
        continue

    year = df.index[0].year
    ausday = df.loc[df.index.date == pd.Timestamp(year=year, month=1, day=26).date(), first_substation]

    if len(ausday) > 0:
        ausday_data[year] = ausday
```

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import pandas as pd

plt.figure(figsize=(12, 6))

for year, series in ausday_data.items():
    # Convert timestamps to a template day (2000-01-26)
    template_index = pd.to_datetime(
        "2000-01-26 " + series.index.strftime("%H:%M:%S")
    )

    plt.plot(template_index, series.values, label=str(year))

# Formatting
ax = plt.gca()
ax.xaxis.set_major_locator(mdates.HourLocator(interval=1))
ax.xaxis.set_major_formatter(mdates.DateFormatter("%H:%M"))

plt.setp(ax.get_xticklabels(), rotation=45, ha="right")

plt.title(f"Australia Day Demand – {full_substation_name} Victoria")
plt.xlabel("Time of Day")
plt.ylabel("Demand (MW)")
plt.legend()
plt.tight_layout()
plt.show()
```
