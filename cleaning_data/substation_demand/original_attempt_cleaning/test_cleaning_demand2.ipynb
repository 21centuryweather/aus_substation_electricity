# Practice plotting
- plotting Aus Day 2012 for the first 3 substations in brisbane

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# Slice Australia Day 2012
aus_day = demand.loc["2012-01-26"]

# First 3 substations
substations = demand.columns[:3]

# Lookup: ID → full name
name_lookup = meta.set_index("ID")["Zone Substation Name"].to_dict()

fig, axes = plt.subplots(3, 1, figsize=(12, 10), sharex=True)

for ax, sub in zip(axes, substations):
    ax.plot(aus_day.index, aus_day[sub])

    full_name = name_lookup.get(sub, sub)  # fallback to ID if missing

    ax.set_title(f"{full_name} ({sub}) — Australia Day 2012")
    ax.set_ylabel("Demand (MW)")

    # Hourly ticks
    ax.xaxis.set_major_locator(mdates.HourLocator(interval=1))
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%H:%M"))

    # Rotate labels
    plt.setp(ax.get_xticklabels(), rotation=45, ha="right")

axes[-1].set_xlabel("Time of Day")

plt.tight_layout()
plt.close()
```
