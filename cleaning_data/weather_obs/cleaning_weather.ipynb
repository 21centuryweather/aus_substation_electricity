## Extracting metro station IDs using available lat/lon

```python
import pandas as pd

df = pd.read_excel("/g/data/ng72/pv3484/substation_data/raw_data/All_stations_information.xlsx")

df["Station_number"] = df["Station_number"].astype(str)

# Melbourne bounding box
melb = df[
    (df["Latitude"].between(-38.5, -37.4)) &
    (df["Longitude"].between(144.4, 145.6))
]

# Brisbane bounding box
bris = df[
    (df["Latitude"].between(-28.4, -27.0)) &
    (df["Longitude"].between(152.5, 153.5))
]

melb_ids = set(melb["Station_number"])
bris_ids = set(bris["Station_number"])

melb_ids, bris_ids
```
